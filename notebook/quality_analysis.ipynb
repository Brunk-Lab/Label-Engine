```python
import pandas as pd

quality_file = '/playpen-raid2/qinliu/data/ecDNA/datasets/quality_val_0422_2024.csv'
df = pd.read_csv(quality_file)
```

```python
count = 0
for index, row in df.iterrows():
    score = row['score']
    name = row['name']
    if name.startswith('NCIH2170') and score >= 3.:
        count += 1
    
print(count)
```
