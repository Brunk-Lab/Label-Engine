```python
import numpy as np
import matplotlib.pyplot as plt

# Define the properties of the Gaussians
mean1 = 0
mean2 = 2.0
variance = 1
sigma = np.sqrt(variance)

# Define the x values
x = np.linspace(-4, 6, 1000)

# Calculate the y values for each Gaussian
gaussian1 = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean1) / sigma) ** 2)
gaussian2 = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean2) / sigma) ** 2)

# Calculate the accumulated values in the overlap region
accumulated_values = gaussian1 + gaussian2

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(x, gaussian1, label=f'Gaussian 1 (mean={mean1})')
plt.plot(x, gaussian2, label=f'Gaussian 2 (mean={mean2})')
plt.plot(x, accumulated_values, label='Accumulated Values', linestyle='--')
plt.fill_between(x, 0, accumulated_values, color='gray', alpha=0.2)
plt.title('Two Gaussians with Accumulation in Overlap Region')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```
