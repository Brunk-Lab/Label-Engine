```python
from tqdm import tqdm
import numpy as np

dataset_folder = '/playpen-raid2/qinliu/data/ecDNA/datasets'
train_file = f'{dataset_folder}/train_0422_2024.txt'
val_file = f'{dataset_folder}/val_0422_2024.txt'

trainset = set()
with open(train_file) as train_f:
    for line in train_f:
        name = line.strip().split('.')[0]
        trainset.add(name)

valset = set()
with open(val_file) as val_f:
    for line in val_f:
        name = line.strip().split('.')[0]
        valset.add(name)

coords_folder = '/playpen-raid2/qinliu/data/ecDNA/coords'
min_distances = []
for elm in tqdm(valset):
    coords = np.load(f'{coords_folder}/{elm}.npy')
    min_distance = 1000000
    for i in range(len(coords) - 1):
        for j in range(i+1, len(coords)):
            distance = np.sqrt(np.sum((coords[i] - coords[j])**2))
            if distance < min_distance:
                min_distance = distance
    min_distances.append(min_distance)
```
