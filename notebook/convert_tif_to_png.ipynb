```python
import os
import numpy as np
from PIL import Image
import imageio
from roifile import ImagejRoi as R
from tqdm import tqdm


img_folder = '/playpen-raid2/qinliu/data/ecDNA/images'
roi_folder = '/playpen-raid2/qinliu/data/ecDNA/ROIs'
tif_folder = '/playpen-raid2/qinliu/data/ecDNA/FISH_Images/Validation_set_manual_mask_binary'
tif_files = sorted(os.listdir(tif_folder))
img_files = sorted(os.listdir(img_folder))

img_file_names = set()
for img_file in img_files:
    img_file_names.add(img_file.strip().split('.')[0])

tif_file_names = set()
for tif_file in tif_files:
    tif_file_names.add(tif_file.strip().split('.')[0])

assert tif_file_names <= img_file_names

tif_file_names = list(tif_file_names)
for i in tqdm(range(len(tif_file_names))):
    file_name = tif_file_names[i]
    file_path = os.path.join(tif_folder, f'{file_name}.tif')
    img_gray = np.array(Image.open(file_path))

    unique_values = np.unique(img_gray)
    assert len(unique_values) == 2 and \
        unique_values[0] == 0 and unique_values[1] == 255
    imageio.imwrite(f'{roi_folder}//{file_name}.png', img_gray)
```
